```python
class Account:
    """Creating an Account class with methods"""
    def __init__(self, balance, interest_rate, initial_interest=0): # Added interest_rate and initial_interest as parameters
        self.balance = balance
        self.interest_rate = interest_rate # Assign interest_rate to the instance
        self.interest = initial_interest # Assign initial_interest to the instance

    # This method sets the balance of the account.
    def set_balance(self, balance):
        """Sets the balance for the for the account"""
        self.balance = balance

    # The method sets the interest gained for the account.
    def set_interest(self, interest):
        """Sets the interest gained for the the account"""
        self.interest = interest

    # This method to update the balance to add interest earned.
    def update_balance(self, interest_earned):
        """Updates the account balance by adding the interest earned."""
        self.balance += interest_earned  # Add the interest to the current balance
        return self.balance # return the updated balance
```
